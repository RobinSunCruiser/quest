# Initialize the QUEST Client
Create an instance of the QuestClient class, configuring server URL, token storage, verbosity, and auto-login settings. Demonstrate how to connect to the server and handle authentication.

```python
# Initialize the QUEST Client

# Create an instance of the QuestClient class
client = QuestClient(
    server_url="http://localhost:3000",  # URL of the QUEST server
    verbose=True,  # Enable verbose logging for detailed output
    auto_login=False  # Automatically prompt for login if no valid token exists
)

# Display client connection status
print(f"Connected to server: {client.connected}")
print(f"Server available: {client.server_available}")

# Set your username and password
username = "user"  # Replace with your username
password = "user"  # Replace with your password

# Login manually
login_success = client.login(username, password)
print(f"Login successful: {login_success}")

# Connect to the server after authentication
if not client.connected:
    connection_success = client.connect()
    print(f"Connection successful: {connection_success}")
```

# List Available Models
Use the list_models() method to retrieve and display the available language models from the QUEST server. Show how to parse and access model information.

```python
# List Available Models

# Retrieve the list of available models from the QUEST server
print("\nRetrieving available models...")
models = client.list_models()

# Check if any models were retrieved
if not models:
    print("No models available or couldn't retrieve model list.")
else:
    # Display the retrieved models
    print(f"\nFound {len(models)} available models:")
    for model in models:
        model_id = model.get("id", "unknown")
        print(f"{model_id}")
```

# Send Requests to Models
Demonstrate how to send non-streaming requests to models using request_completion(). Show how to specify the model ID, prompt, system instructions, and model-specific options.

```python
# Send Requests to Models

# Select a model ID from the list of available models
if models:
    selected_model_id = "perplexity-r1-1776"  # Use the first model in the list
    print(f"\nUsing model: {selected_model_id}")

    # Define the prompt and system instructions
    prompt = "What is the capital of France?"
    system_prompt = "You are a geography expert. Provide brief answers."

    # Define model-specific options
    options = {
        "temperature": 0.3,  # Control randomness of the outpu
    }

    # Send a non-streaming request to the selected model
    print("\nSending a completion request...")
    response = client.request_completion(
        model_id=selected_model_id,
        prompt=prompt,
        system_prompt=system_prompt,
        options=options,
    )

    # Extract and display the response content
    content = client.get_response_content(response)
    print(f"\nResponse to '{prompt}':\n{content}")
else:
    print("\nNo models available to send requests.")
```

# Evaluate Model Responses
Demonstrate the evaluate_responses() method to compare and analyze responses from multiple models. Show how to interpret the evaluation metrics and results.

```python
# Evaluate Model Responses

# Define responses from multiple models for evaluation
responses = [
    {
        "modelID": "openai-gpt-3.5-turbo",
        "message": "Paris is the capital of France.",
    },
    {
        "modelID": "ollama-llama2",
        "message": "The capital city of France is Paris.",
    },
]

# Define evaluation options
evaluation_options = {
    "removeThinkingBlock": True,  # Remove "thinking" sections before evaluation

}

# Send evaluation request to the QUEST server
print("\nEvaluating responses from multiple models...")
evaluation_results = client.evaluate_responses(
    model_responses=responses,
    options=evaluation_options,
    timeout=30,  # Set timeout for the evaluation request
)

# Check if evaluation was successful
if "error" in evaluation_results:
    print(f"Evaluation error: {evaluation_results['error']}")
else:
    # Display the evaluation results
    print("\nEvaluation results:")
    print(json.dumps(evaluation_results, indent=2))
```
